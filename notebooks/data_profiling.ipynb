```python
import pandas as pd

# 1. Cargar el dataset sin filtros
file_path = "../data/raw/Registro_Nacional_de_Turismo_-_RNT_20260915.csv"
df = pd.read_csv(file_path, low_memory=False)

# 2. Resumen general de filas y columnas
print("==========================================")
print(f"Total de Registros (Filas): {len(df):,}")
print(f"Total de Atributos (Columnas): {len(df.columns)}")
print("==========================================")

# 3. Inspección de tipos de datos y valores nulos
profile_df = pd.DataFrame({
    'Tipo_Dato': df.dtypes.astype(str),
    'Valores_Nulos': df.isnull().sum(),
    'Porcentaje_Nulos (%)': (df.isnull().sum() / len(df)) * 100,
    'Valores_Unicos': df.nunique()
})

profile_df
```

```python
# 1. Estadísticas descriptivas de las métricas (Rangos, promedios, mínimos y máximos)
metrics = ['NUMERO_DE_HABITACIONES', 'NUMERO_DE_CAMAS', 'NUMERO_DE_EMPLEADOS']
print("=== ESTADÍSTICAS DESCRIPTIVAS DE MÉTRICAS ===")
display(df[metrics].describe().T[['min', 'mean', '50%', 'max', 'std']])

# 2. Verificación de reglas de calidad e inconsistencias (Valores negativos)
print("\n=== VERIFICACIÓN DE VALORES NEGATIVOS ===")
for col in metrics:
    negatives = (df[col] < 0).sum()
    print(f"Valores negativos en {col}: {negatives}")

# 3. Cobertura temporal (Años únicos registrados)
print("\n=== COBERTURA TEMPORAL (AÑOS) ===")
print(sorted(df['AÑO'].unique()))
```
